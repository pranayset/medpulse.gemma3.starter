```python
from pathlib import Path
import json, sys
ROOT = Path.cwd()
if not (ROOT / 'data').exists() and (ROOT.parent / 'data').exists():
    ROOT = ROOT.parent
print('Project root:', ROOT.resolve())
print('Python:', sys.version)
```

## 1. Load synthetic examples


```python
sample_path = ROOT / 'data' / 'sample' / 'synthetic_discharge_examples.jsonl'
records = [json.loads(line) for line in sample_path.read_text(encoding='utf-8').splitlines() if line.strip()]
assert records and all({'instruction', 'response'} <= set(r) for r in records)
print(f'Loaded {len(records)} synthetic examples')
print(records[0]['instruction'])
```

## 2. Safety checklist before using any model
- Use synthetic, non-identifiable examples only.
- Never paste Hugging Face tokens or credentials into cells.
- Do not download or commit trained adapters/model weights to this repository.
- Clear notebook outputs before publishing.
- Verify generated text with a qualified clinician; do not use it for care decisions.


```python
def safe_rewrite_prompt(text: str) -> str:
    if not isinstance(text, str) or not text.strip():
        raise ValueError('Provide non-empty synthetic text only')
    return ('Educational synthetic-data exercise. Do not invent clinical facts. '
            'This output is not medical advice and requires qualified human review.\n\n'
            + text.strip())

print(safe_rewrite_prompt(records[0]['instruction']))
```
